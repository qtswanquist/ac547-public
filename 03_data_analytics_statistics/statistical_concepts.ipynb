## Basic statistics – location
We will begin by discussing statistics that are useful for describing the location of data. We will start with a very simple example. Below, you will find a listing of college football coaches who led a program at some point in the past five seasons (2021 to 2025) ordered by the number of total titles. 

| Coach     | Titles |
|-----------|--------|
| Saban     | 7      |
| Swinney   | 2      |
| Smart     | 2      |
| Harbaugh  | 1      |
| Brown     | 1      |
| Day       | 1      |
| Cignetti  | 1      |

A few other coaches from this period also have one title; we list seven here to keep the example small. Throughout this section, $N = 7$ refers to the number of coaches and $x_i$ refers to the number of titles won by coach $i$. For this example, we will consider this the population.

### Mean or average
- Formula:

$$
\mu = \frac{1}{N}\sum_{i=1}^{N} x_i
       = \frac{x_1 + x_2 + \cdots + x_N}{N}
       = \frac{7 + 2 + 2 + 1 + 1 + 1 + 1}{7}
       = \frac{15}{7}
       \approx 2.14
$$

- The population mean (represented as $\mu$) is the most commonly used measure of location.
- Note: The mean is disproportionately affected by outliers (extreme values). Notice that no coach in the table has actually won 2.14 titles. Saban's number of titles has a large effect on the mean in this example.

---

### Percentile
- A percentile describes the relative position of a value in an ordered dataset. The 50th percentile is the **median**.

#### The median (50th percentile)
- Represents the “middle” point in the data. To find it, sort the values from smallest to largest and take the middle one. With an even number of observations, average the two middle values.
- Sorting our seven coaches by titles gives:

$$
1,\; 1,\; 1,\; \mathbf{1},\; 2,\; 2,\; 7
$$

- With $N = 7$, the middle observation is the 4th, so the median is **1**. 
- Relative to the mean, the median is less affected by outliers. If Saban had won 70 titles instead of 7, the mean would jump to roughly 11.1, but the median would not budge from 1. 

#### Other percentiles
- The same logic extends to other percentiles. The 25th percentile, or first quartile ($Q_1$), is the value that separates the lowest 25% of observations from the rest. The 75th percentile, or third quartile ($Q_3$), separates the lowest 75% of observations from the highest 25%.
- Percentiles are more informative in larger datasets. With only seven coaches, four of whom are tied at a single title, the 25th and 50th percentiles both land on 1, so the quartiles have little to tell us here. We will get much more out of percentiles once we turn to the diamonds data. For numeric values, `pandas` reports the 25th, 50th, and 75th percentiles as the `25%`, `50%`, and `75%` rows every time we call `.describe()`, and we will use $Q_1$ and $Q_3$ again in the next section to define the interquartile range.

---

### Mode
- The most commonly occurring value in the data.
- To find it, count how often each value appears:

| Titles | Coaches                        | Count |
|--------|--------------------------------|-------|
| 1      | Harbaugh, Brown, Day, Cignetti | 4     |
| 2      | Swinney, Smart                 | 2     |
| 7      | Saban                          | 1     |

- One title is the most common outcome, so the mode is **1**.
- Unlike the mean and median, a dataset can have more than one mode (when two or more values tie) or no useful mode at all (when every value appears exactly once). The mode is most informative for categorical data, or for numeric data with repeated values.

---

### Comparing the measures
For these coaches, the mean (2.14) sits well above both the median (1) and the mode (1). Whenever the mean is pulled to the right of the median like this, it is a signal that a few large values are stretching the data in that direction. We will come back to this idea when we discuss skewness in the **Distributions** section below.

## Basic statistics – variability
We will now turn to basic statistics that are useful for describing the variability of data, or how spread out the values are around the center. 

### Variance
- The average of the squared differences from the mean.
- Formula:

$$
\sigma^2 = \frac{1}{N}\sum_{i=1}^{N}(x_i - \mu)^2
         = \frac{(x_1-\mu)^2 + (x_2-\mu)^2 + \cdots + (x_N-\mu)^2}{N}
$$

Using the example data, where $\mu \approx 2.14$:

$$
\sigma^2
= \frac{(7-2.14)^2 + (2-2.14)^2 + (2-2.14)^2 + (1-2.14)^2 + (1-2.14)^2 + (1-2.14)^2 + (1-2.14)^2}{7}
\approx 4.12
$$

- The calculation above represents the population variance. Note that the variance is also disproportionately affected by outliers (extreme values). Saban alone contributes about 23.6 of the 28.9 total squared deviation, or roughly 82 percent of the variance, from a single observation.

---

### Standard deviation
- The square root of the variance.
- Formula:

$$
\sigma = \sqrt{\sigma^2} = \sqrt{4.12} \approx 2.03
$$

- Like the variance, the standard deviation measures how spread out the data is and is also affected by outliers.
- The standard deviation is often easier to interpret than the variance because it is expressed in the same units as the original data (here, national titles).

---

### Interquartile range (IQR)
- The difference between the 75th and 25th percentiles. Represents the range of the middle 50 percent of the data.
- Formula:

$$
\text{IQR} = Q_3 - Q_1
$$

- Relative to the variance and the standard deviation, the interquartile range is less affected by outliers. Because the IQR depends only on \(Q_1\) and \(Q_3\), extreme tail values usually have little or no effect on it. If Saban had won 70 titles instead of 7, the variance would balloon from about 4.12 to roughly 577 and the standard deviation from about 2.03 to roughly 24, but the IQR would not move.
- As with the other percentiles, seven coaches are too few for this measure to be interesting. We will come back to the IQR when we read the `25%` and `75%` rows of `.describe()` on the diamonds data.

## Distributions

Location and variability statistics are summaries of something larger: the **distribution** of the data, which describes how values are spread across the range of possible outcomes. Two datasets can share a mean and a standard deviation and still look nothing alike, so it is worth looking at the whole shape rather than just the summary numbers.

The illustrations are built with Claude Code. Don't worry too much about the code here... rather this is intended to help you understand the concepts in the course.

### The normal distribution

The normal distribution is symmetric around its center, which means the mean and median land in the same place. Most observations sit near the middle and become progressively rarer as you move into either tail. A useful rule of thumb here is that roughly 68 percent of observations fall within one standard deviation of the mean, 95 percent within two, and 99.7 percent within three.

```python
# Build a standard normal curve (mean of 0, standard deviation of 1)
x_values = np.linspace(-4, 4, 1000) # Create 1,000 evenly spaced points from -4 to 4
density = norm.pdf(x_values) # Height of the normal curve at each point

sns.lineplot(x=x_values, y=density, color='black') # Draw the curve

# Shade the regions within 1, 2, and 3 standard deviations of the mean
regions = [(3, 0.20, '99.7%'), (2, 0.30, '95%'), (1, 0.40, '68%')]

for num_sd, shade, share in regions: # Loop through each region
    plt.fill_between(x_values,
                     density,
                     where=np.abs(x_values) <= num_sd,
                     color='steelblue',
                     alpha=shade,
                     label=f'Within {num_sd} SD ({share})'
                     ) # Shade the area under the curve

plt.axvline(0, color='red', linestyle='--', label='Mean') # Mark the mean

plt.xlabel('Standard Deviations from the Mean') # Add x-axis label
plt.ylabel('Probability Density') # Add y-axis label
plt.title('The Normal Distribution') # Add a title to the plot
plt.legend() # Add a legend

plt.show() # Display the plot
```

### Skewed distributions

When data are symmetric, the mean and median tend to line up in the center. When the distribution is skewed, extreme values in the tail pull the mean toward the skew, causing it to move away from the median. The gap between the mean and median can therefore provide a useful clue about asymmetry and the influence of extreme observations.

The three distributions below are built to share the same population mean and population standard deviation, so the only thing that changes from panel to panel is the shape. The bottom panel is the top one reflected, and all three are drawn on the same axes so you can compare them directly. Watch the distance between the mean (dashed red line) and the median (dotted black line) as the skew changes. This is the same pattern we saw with the coaches, where Saban's 7 titles pulled the mean above the median.

```python
np.random.seed(42) # Set a random seed for reproducibility

# Build three distributions that share the same mean (2) and the same
# standard deviation (about 1.41), so that only the shape differs
right_skew = np.random.gamma(shape=2, scale=1, size=10000)
symmetric = np.random.normal(loc=2, scale=np.sqrt(2), size=10000)
left_skew = 4 - right_skew # Reflecting the same draw gives an exact mirror

# Stack the three arrays into one tidy DataFrame with a label column
skew_shapes = pd.concat([
    pd.DataFrame({'shape': 'Positive (Right) Skew', 'value': right_skew}),
    pd.DataFrame({'shape': 'Symmetric', 'value': symmetric}),
    pd.DataFrame({'shape': 'Negative (Left) Skew', 'value': left_skew}),
], ignore_index=True)

shape_order = skew_shapes['shape'].unique() # Panel order, as stacked above

# Seaborn draws one panel per shape, on shared axes and shared bins
grid = sns.displot(data=skew_shapes,
                   x='value',
                   row='shape',
                   row_order=shape_order,
                   stat='density',
                   common_norm=False, # Scale each panel on its own
                   bins=np.linspace(-6, 10, 65),
                   height=2.5,
                   aspect=3
                   )

# Mark the mean (dashed red) and median (dotted black) in each panel
center = skew_shapes.groupby('shape')['value'].agg(['mean', 'median'])

for ax, shape_name in zip(grid.axes.flat, shape_order):
    mean_value, median_value = center.loc[shape_name]
    ax.axvline(mean_value, color='red', linestyle='--',
               label=f'Mean = {mean_value:.2f}')
    ax.axvline(median_value, color='black', linestyle=':',
               label=f'Median = {median_value:.2f}')
    ax.set_title(shape_name)
    ax.legend()

grid.set_axis_labels('Value', 'Density') # Label the axes
grid.figure.suptitle('How Skewness Moves the Mean Away from the Median',
                     y=1.02
                     ) # Add an overall title

plt.show() # Display the plot
```

Notice the pattern in the three panels above:

- **Right skew:** the long tail is on the high side, so the mean sits *above* the median.
- **Symmetric:** the mean and median sit on top of each other.
- **Left skew:** the long tail is on the low side, so the mean sits *below* the median.

This gives you a quick diagnostic. Any time you call `.describe()` and see the mean sitting well away from the 50th percentile, you have a clue about the shape of the data before plotting anything.

## Sampling: why would we ever rely on a sample?

Often, we don’t have the resources to collect data for the entire population. Similarly, we may not have access to the entire population. However, a subset of data, when evaluated properly, can be used to make inferences about a population, but we must be smart about it!

Below, we will learn about two very important statistical principles that underlie our ability to draw inferences from a sample.

### Central limit theorem and the law of large numbers

- The **central limit theorem** tells us that the means from samples have an approximately normal distribution (when each sample size is sufficiently large).
- The **law of large numbers** tells us that as we increase sample size, we tend to get closer to the true average outcome.
- Together, these concepts allow us to infer that our **sample** tells us something meaningful about the **population**.

**Free references:**
- [Central Limit Theorem (CLT): Definition and Key Characteristics](https://www.investopedia.com/terms/c/central_limit_theorem.asp)
- [Law of Large Numbers – Wikipedia](https://en.wikipedia.org/wiki/Law_of_large_numbers)

### Sampling simulation
Let’s demonstrate these concepts using a sample of coin flips from a fair coin. After that, we will simulate sampling from the diamonds dataset to see how sample means behave when the underlying population is not normally distributed.

#### Simulate and plot data on coin flips
Imagine that 1,000 students on campus are each given a fair coin (equal chance of heads or tails). Each student is asked to flip the coin 10 times and record the number of heads and tails (0=tails and 1=heads). In expectation, half of the flips should be tails and half of the flips should be heads. Each coin flip is an independent trial. In other words, the outcome of one flip does not affect the outcome of the next flip. This means that each flip has a 50% chance of being heads and a 50% chance of being tails, regardless of the outcomes of previous flips (if you have ever thought that a coin is "due" for heads or tails, you are falling victim to the gambler's fallacy!).

We are going to simulate a scenario where many people flip a coin several times. Let’s store the number of coin flips per person in `n_tries` and the number of people flipping coins in `n_samples`.

```python
n_tries = 10
n_samples = 1000
```

Next, we will build a simulation using a “for loop” where each person in `n_samples` flips a coin `n_tries` times. Each flip represents a random draw from values 0 or 1 (where 0 represents tails and 1 represents heads). The results of each sample are stored in a list called `final_heads`.

```python
np.random.seed(42) # Set a random seed for reproducibility

final_heads = [] # Initialize an empty list to store the number of heads for each sample

for _ in range(n_samples): # Loop through each sample
    heads = [0] # Start with zero heads for each sample
    for _ in range(n_tries): # Loop through each flip
        coin = np.random.randint(0, 2) # Randomly select a coin flip outcome (0 or 1)
        heads.append(heads[-1] + coin) # Add the result of the flip to the heads list
    final_heads.append(heads[-1]) # Append the total number of heads for this sample to final_heads

final_heads = pd.DataFrame(final_heads, columns=['heads']) # Convert final_heads to a DataFrame for easier handling

final_heads = final_heads.groupby('heads').size().reset_index(name='count') # Aggregate the data so that we have the count of heads for each sample

print(final_heads)
```

Inspect the output from the loop above. For each possible number of heads (from 0 to `n_tries`), we have a count of how many samples had that number of heads. Next, let's plot the distribution of the number of heads using a bar plot.

```python
sns.barplot(x='heads',
            y='count',
            data=final_heads
            ) # Create a bar plot of the number of heads across samples

plt.xlabel(f'Heads from {n_tries} Coin Flips') # Add x-axis label
plt.ylabel('Count') # Add y-axis label
plt.title(f'Distribution of Heads Across {n_samples} Samples') # Add a title to the plot

plt.show()
```

Notice that the most common number of heads for each coin flipper falls around 5, but many (most!) are not exactly 5. In fact, some coin flippers are far away from 5 (e.g., 2 tails and 8 heads in 10 flips). However, as we get farther away from the “expectation” of 5, the instances become more rare. The distribution is discrete, but it resembles a normal distribution. This foreshadows the central limit theorem... which we will see in more detail in exercises 1-2 at the end of this module.

## Sample bias

Ideally, our **samples** are *representative* of the **population** or group we want to study. When a sample is **not representative** of the target population, this is known as **sample (selection) bias**.

Common types of sample selection bias include:
  - **Self-selection bias:** Individuals (e.g., voters or survey respondents) choose whether or not to participate, which can skew results.
  - **Survivorship bias:** Only “successful” or observable outcomes are included, while “unsuccessful” or missing observations are ignored.

## Simulate sampling from the diamonds dataset

The coin flip example may seem a little abstract. Let’s explore these principles in action using the diamonds dataset we loaded earlier. Here, let’s imagine that 1,000 auditors each take an independent sample from the diamonds dataset. Each auditor takes a sample of 30 diamonds and is asked to determine the average carat weight.

As above, let’s store our number of samples, sample size, and other information in objects (so that we can make easy changes later).

```python
n_samples = 1000 # Number of samples to draw
sample_size = 30 # Number of diamonds in each sample
sample_variable = 'carat' # The variable we are averaging
```

Next, let’s take a look at summary statistics and the distribution of carat weight. Note that we use `diamonds` here rather than `diamonds_small`, so that we are sampling from the full diamonds dataset.

```python
print(diamonds[[sample_variable]].describe()) # Print summary statistics for the sampled variable
```

Next, let's visualize this distribution with a histogram.

```python
sns.histplot(data=diamonds,
             x=sample_variable,
             bins=50
             )

plt.title(f'Distribution of {sample_variable}')
plt.xlabel(sample_variable)
plt.ylabel('Count')

plt.show()
```

Notice that the distribution of carat weight is positively skewed and has several breakpoints. Importantly, it is **not** normally distributed. Keep that in mind when we plot the sample averages below.

Let’s take our own sample of the diamonds dataset.

```python
example_sample = diamonds.sample(n=sample_size, replace=False, random_state=42) # Draw one sample of diamonds

print(example_sample[[sample_variable]].describe()) # Display summary statistics of the sample
```

Take a look at your sample above. Change the options in `sample()` and see what changes.

Following the coin flips example, let’s simulate taking independent samples, determining the sample average, and plotting the results.

```python
np.random.seed(42) # Set a random seed for reproducibility

samples_list = [] # Initialize an empty list to store each sample

for s in range(n_samples): # Loop through each sample
    sample_df = diamonds.sample(n=sample_size, replace=False)[[sample_variable]].copy() # Take a sample of diamonds
    sample_df['sample'] = s # Add a sample identifier
    samples_list.append(sample_df) # Append the sample DataFrame to the list

diamonds_samples = pd.concat(samples_list, ignore_index=True) # Combine all samples into a single DataFrame

diamond_averages = (
    diamonds_samples
    .groupby('sample') # Group by the sample identifier
    .agg(average_value=(sample_variable, 'mean')) # Average the variable within each sample
    .reset_index() # Reset the index and make sample a column again
)

diamond_averages.head() # Display the first few rows of the diamond averages DataFrame
```

```python
sns.histplot(diamond_averages['average_value'],
             bins=30,
             color='blue',
             edgecolor='black',
             alpha=0.7
             )

plt.title(f'Distribution of Sample Means for {sample_variable}')
plt.xlabel('Sample Mean')
plt.ylabel('Count')

plt.show()
```
