## Data analytics exercise

Next, you will complete a short data analytics exercise to make sure you are up and running. Try to follow along, but don't worry if you get stuck or don't understand. We will do a deeper dive into the topics later in the course. If you can get the code below to run, you are good to go for the course!

We will use a classic teaching dataset called **Anscombe's Quartet**, built by the statistician Francis Anscombe in 1973. It is four small datasets that share nearly identical summary statistics but look nothing alike once you plot them. We come back to it in the regression module, where that turns out to matter a great deal. Today we are only using it to confirm that everything on your machine works.

Let's load the data with `pd.read_pickle` from a file that sits in this folder, right next to the notebook. Throughout this course, the data you need will live alongside the notebook that uses it.

````python
anscombe = pd.read_pickle('anscombe.pkl') # Load Anscombe's dataset from a local pickle file

anscombe.sample(10, random_state=42)  # Display a random sample of 10 rows from the dataset
````

Take a look at the sample from the dataset above. Notice that it contains four different datasets (i.e., **I**, **II**, **III**, and **IV**) with two variables each (i.e., `x` and `y`). Each dataset has a different relation between **x** and **y**. The goal of this exercise is to visualize the data, perform a regression analysis, and output the predicted values and residuals. Let's do some data wrangling to select only the first dataset (i.e., **I**) and keep only the `x` and `y` variables.

````python
anscombe_i = anscombe[anscombe['dataset'] == 'I']  # Select only dataset I
anscombe_i = anscombe_i[['x', 'y']]  # Select only x and y columns
anscombe_i # Display the filtered dataset
````

## Visualizing the data

Take another look at the dataset above. Notice that we have two separate variables (i.e., `x` and `y`) that contain a list of numbers. To visualize, let's create a scatter plot of **y** on **x** including a regression line. Again, notice that we use **plt.** and **sns.** to refer to matplotlib and seaborn imported above.

You should get a chart with eleven points and a red line running through them. That line is the "line of best fit." In the next section, we calculate that same line numerically using regression.

````python
sns.regplot(data=anscombe_i,
            x='x',
            y='y',
            ci=None,
            line_kws={'color': 'red'}
            ) # Create a scatter plot with a regression line

plt.title("Anscombe's Quartet - Dataset I") # Add a title to the plot

plt.show() # Display the plot
````

## Regression

Next, we will perform a simple linear regression to predict **y** using **x**. In regression parlance, we call this “a regression of **y** on **x**.” This is the same line you saw on the chart above, worked out numerically.

The output below has a lot in it, and most of it will not mean anything to you yet. We will learn more about this later in the course. If you look at one thing, find `Intercept` and `x` in the middle table: those two numbers are the red line's starting height and its steepness (remember `y = mx + b` from middle school?).

````python
model = smf.ols('y ~ x', data=anscombe_i).fit() # Fit a linear regression model using statsmodels
print(model.summary()) # Display the summary of the regression model
````

## Output predictions and residuals

Finally, we output the predicted (fitted) values and the residuals for each observation. The **fitted** value is what the line predicts `y` should be at a given level of `x`. The **residual** is how far off each prediction was or the vertical gap between the actual point and the red line.

````python
anscombe_i['fitted'] = model.fittedvalues # Add fitted values to the dataframe
anscombe_i['resid'] = model.resid # Add residuals to the dataframe
anscombe_i # Display the updated dataframe
````
